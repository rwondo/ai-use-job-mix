```python
import pandas as pd
df = pd.read_csv("../data/raw/anthropic/aei_claude_ai_2026-06-26.csv",
                 dtype={"node_external_id": str}, low_memory=False)
print(len(df))
for col in ["geo_level", "category_name", "hierarchy_level", "metric_id", "date_start"]:
    print(col, df[col].unique()[:20])
```

```python
it = df[
    (df["geo_id"] == "ITA") &
    (df["geo_level"] == "country") &
    (df["category_name"] == "soc_occupation") &
    (df["hierarchy_level"] == 0) &
    (df["metric_id"] == "pct") &
    (df["date_start"] == "2026-04-01")
]
print(len(it))
print(it.sort_values("value", ascending=False)[["node_name", "node_external_id", "value"]].head(10))
print("Coverage:", it["value"].sum())
```

```python
c = df[df["geo_level"] == "country"]
print("Country rows:", len(c))
print("Country codes:", c["geo_id"].unique()[:40])

s = c[c["category_name"] == "soc_occupation"]
print("Job rows:", len(s))
print("Levels:", s["hierarchy_level"].unique())
print("Metrics:", s["metric_id"].unique())
```

```python
europe = ["AUT","BEL","BGR","HRV","CYP","CZE","DNK","EST","FIN","FRA","DEU","GRC",
          "HUN","IRL","ITA","LVA","LTU","LUX","MLT","NLD","POL","PRT","ROU","SVK",
          "SVN","ESP","SWE","NOR","ISL","CHE","SRB","MKD","TUR","BIH"]
jobs = df[
    (df["geo_level"] == "country") &
    (df["category_name"] == "soc_occupation") &
    (df["hierarchy_level"] == 0) &
    (df["metric_id"] == "pct") &
    (df["geo_id"].isin(europe))
]
coverage = jobs.groupby(["geo_id", "date_start"])["value"].sum().unstack()
coverage = coverage.sort_values("2026-04-01")
print(coverage)
coverage.to_csv("../output/tables/coverage.csv")
```
